```python
import pandas as pd
data = {
    'name': ['Анна', 'Борис', 'Вика', 'Денис', 'Елена'],
    'age': [21, 22, 23, 24, 15],
    'city': ['Москва', 'Санкт-Петербург', 'Казань', 'Москва', 'Сочи'],
    'math_grade': [85, 92, 88, 78, 95],
    'physics_grade': [90, 89, 85, 88, 92]
}

df = pd.DataFrame(data)
# Посмотри на первые 3 строки датафрейма с помощью head().
df.head(3)
```

```python
# Посмотри на последние 2 строки датафрейма с помощью tail().
df.tail(2)
```

```python
# Получите информацию о DataFrame, чтобы проверить типы данных и другие свойства с помощью info().
df.info()
```

```python
# 5. Добавление нового столбца average_grade
df['average_grade'] = (df['math_grade'] + df['physics_grade']) / 2
df.head()
```

```python
# Отфильтруй студентов, у которых возраст больше 22 лет, и выведи результат.
df_filtered = df[df['age'] > 22]
df_filtered
```

```python
df['isPassed'] = (df['average_grade'] >= 90)
df['isPassed']
```

```python
df.to_csv('data.csv', index=False)
df
```
